```python
import numpy as np

np.random.seed(42)
# Student Performance dataset (marks out of 100) + 2 outliers
marks = np.append(np.random.normal(65, 10, 48).round(), [5, 100])

mean   = np.mean(marks)
median = np.median(marks)
std    = np.std(marks)
mn, mx = np.min(marks), np.max(marks)

print(f"Mean: {mean:.2f}\nMedian: {median:.2f}")
print(f"Std Dev: {std:.2f}\nMin: {mn}\nMax: {mx}")

# Outliers (z-score > 2)
z = (marks - mean) / std
print("Outliers:", marks[np.abs(z) > 2])
```
